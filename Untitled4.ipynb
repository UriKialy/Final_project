```python
# %% Locate or build labels CSV
def gather_images(root: Path):
    IMG_EXTS = {'.dcm', '.png', '.jpg', '.jpeg'}
    files = [p for p in root.rglob('*') if p.is_file() and p.suffix.lower() in IMG_EXTS]
    return files

def index_by_basename(files):
    # map basename -> first full path (handles Windows-origin CSVs)
    idx = {}
    for p in files:
        idx.setdefault(p.name.lower(), p)
    return idx

def find_labels_csv(candidates_roots):
    hits = []
    for base in candidates_roots:
        if not base.exists(): continue
        for p in base.rglob('*.csv'):
            try:
                df = pd.read_csv(p, nrows=5)
                low = [c.lower() for c in df.columns]
                if ('label' in low) and (('dicom_path' in low) or ('path' in low) or ('image' in low)):
                    hits.append(p.resolve())
            except Exception:
                pass
    # prefer names that look right
    hits = sorted(hits, key=lambda p: ('paths' not in p.name.lower(), 'labels' not in p.name.lower(), p.name))
    return hits[0] if hits else None

def normalize_paths_in_df(df, data_root: Path):
    # try to get a single 'path' col and an int 'label' col
    cols = {c.lower(): c for c in df.columns}
    pcol = cols.get('dicom_path') or cols.get('path') or cols.get('image') or list(df.columns)[0]
    lcol = cols.get('label') or list(df.columns)[1]

    paths = df[pcol].astype(str).tolist()
    labels = df[lcol].astype(int).tolist()

    # remap paths to Drive by basename match if necessary
    files = gather_images(data_root)
    by_name = index_by_basename(files)

    fixed = []
    missing = 0
    for s in paths:
        # sanitize
        s2 = s.strip().strip('"').strip("'").replace('\\', '/')
        candidate = Path(s2)
        if candidate.exists():
            fixed.append(candidate.resolve())
            continue
        # if not exists, try join with DATA_ROOT or match by basename
        name = Path(s2).name.lower()
        hit = by_name.get(name, None)
        if hit is not None:
            fixed.append(hit.resolve())
        else:
            # try inside DATA_ROOT directly
            guess = (data_root / name).resolve()
            if guess.exists():
                fixed.append(guess)
            else:
                missing += 1
                fixed.append(None)

    if missing:
        print(f"⚠️  {missing} paths from CSV were not found under Drive; those rows will be dropped.")

    new_df = pd.DataFrame({
        'path': [str(p) for p in fixed if p is not None],
        'label': [labels[i] for i,p in enumerate(fixed) if p is not None]
    })
    return new_df

# Try to find an existing CSV
csv_guess = find_labels_csv([DATA_ROOT, DATA_ROOT.parent])
if csv_guess:
    print("Found labels CSV:", csv_guess)
    df_all = normalize_paths_in_df(pd.read_csv(csv_guess), DATA_ROOT)
else:
    print("No labels CSV found — building a placeholder CSV with label=0. Edit labels if needed.")
    files = gather_images(DATA_ROOT)
    df_all = pd.DataFrame({'path':[str(p) for p in files], 'label':[0]*len(files)})

# sanity
df_all = df_all.drop_duplicates('path').reset_index(drop=True)
print(df_all.head(3))
print("Total rows:", len(df_all))

# Optional: save the normalized CSV next to runs
CSV_OUT = OUT_DIR / 'train_paths_labels.normalized.csv'
df_all.to_csv(CSV_OUT, index=False)
print("Wrote:", CSV_OUT)
```

```python
# %% [colab] Build df_all from Drive (robust DICOM discovery + CSV labels)
!pip -q install --no-input pandas pydicom opencv-python

import os, sys, re, json, glob, random, numpy as np, pandas as pd, cv2, pydicom
from pathlib import Path

IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    from google.colab import drive
    drive.mount('/content/drive', force_remount=True)

# ---- Set your dataset root explicitly (this is the one you said had ~500 imgs)
DATA_ROOT = Path('/content/drive/MyDrive/CBIS-DDSM/CBIS-DDSM').resolve()
if not DATA_ROOT.exists():
    # fallback to the parent you showed in your screenshot
    DATA_ROOT = Path('/content/drive/MyDrive/CBIS-DDSM').resolve()
assert DATA_ROOT.exists(), f"DATA_ROOT not found: {DATA_ROOT}"

OUT_DIR = Path('/content/drive/MyDrive/cbis_cls_runs'); OUT_DIR.mkdir(parents=True, exist_ok=True)
print("DATA_ROOT:", DATA_ROOT)

# ---------- 1) Index all DICOM files (even those with no extension) ----------
def list_candidate_files(root: Path):
    # everything except tiny side files
    for p in root.rglob('*'):
        if not p.is_file():
            continue
        if p.name.lower() in {'license.txt','readme.txt'}:
            continue
        yield p

def is_dicom_file(p: Path) -> bool:
    # fast path by extension (case-insensitive)
    if p.suffix.lower() in {'.dcm', '.dicom'}:
        return True
    # try to parse the header without pixel data
    try:
        ds = pydicom.dcmread(str(p), stop_before_pixels=True, force=True)
        # must at least look like DICOM (has FileMeta or a few standard tags)
        return hasattr(ds, 'SOPClassUID') or hasattr(ds, 'PatientID') or hasattr(ds, 'StudyInstanceUID')
    except Exception:
        return False

def get_uids(p: Path):
    siu, sop = None, None
    try:
        ds = pydicom.dcmread(str(p), stop_before_pixels=True, force=True)
        siu = getattr(ds, 'SeriesInstanceUID', None)
        sop = getattr(ds, 'SOPInstanceUID', None)
    except Exception:
        pass
    return siu, sop

print("Indexing DICOM files … (this will walk the CBIS-DDSM tree)")
dicom_rows = []
for f in list_candidate_files(DATA_ROOT):
    if is_dicom_file(f):
        siu, sop = get_uids(f)
        dicom_rows.append((str(f.resolve()), siu, sop))
dicom_df = pd.DataFrame(dicom_rows, columns=['path','SeriesInstanceUID','SOPInstanceUID'])
print(f"Indexed DICOM files: {len(dicom_df)}")

# ---------- 2) Load labels from your CSVs ----------
# We will prefer cbis_ddsm_dicom_with_labels.csv, then train_paths_labels.csv, then any CSV with label-ish columns.
CSV_PREF_ORDER = [
    'cbis_ddsm_dicom_with_labels.csv',
    'train_paths_labels.csv',
    'labels.csv'
]

def find_csvs(root: Path):
    # look in root and one level up
    hits = []
    for p in list(root.glob('*.csv')) + list((root/'csv').glob('*.csv')) + list(root.parent.glob('*.csv')):
        hits.append(p.resolve())
    # de-dup & sort by our preference
    uniq = {}
    for p in hits:
        uniq[p.name.lower()+str(p)] = p
    hits = list(uniq.values())
    hits.sort(key=lambda p: (CSV_PREF_ORDER.index(p.name) if p.name in CSV_PREF_ORDER else 9999, p.name))
    return hits

csv_candidates = find_csvs(DATA_ROOT)
print("CSV candidates:")
for i, p in enumerate(csv_candidates[:10], 1):
    print(f" {i}) {p.name} — {p}")

def coerce_label(v):
    sv = str(v).strip().lower()
    if sv in {'1','true','malignant','malign','cancer','pos','positive'}:
        return 1
    if sv in {'0','false','benign','normal','neg','negative'}:
        return 0
    try:
        vi = int(float(sv))
        return 1 if vi==1 else 0
    except:
        return None

def read_label_csv(pcsv: Path):
    try:
        df = pd.read_csv(pcsv)
    except Exception as e:
        print(f"  ⚠️ failed to read {pcsv}: {e}")
        return None

    cols = {c.lower(): c for c in df.columns}
    # detect label column
    lcol = (cols.get('label') or cols.get('labels') or cols.get('benign_malignant') or
            cols.get('pathology') or cols.get('target') or None)
    # detect path-like columns
    pcol = (cols.get('path') or cols.get('dicom_path') or cols.get('image') or
            cols.get('filepath') or cols.get('file') or None)
    # UID columns
    siu_col = (cols.get('seriesinstanceuid') or cols.get('series_instance_uid') or
               cols.get('series_uid') or None)
    sop_col = (cols.get('sopinstanceuid') or cols.get('sop_instance_uid') or
               cols.get('instance_uid') or cols.get('image_uid') or None)

    # unify to a mini frame with any of: path / SeriesInstanceUID / SOPInstanceUID + label
    out = pd.DataFrame()
    if pcol is not None:
        out['path'] = df[pcol].astype(str)
    if siu_col is not None:
        out['SeriesInstanceUID'] = df[siu_col].astype(str)
    if sop_col is not None:
        out['SOPInstanceUID'] = df[sop_col].astype(str)
    if lcol is not None:
        out['label'] = df[lcol].apply(coerce_label)
    else:
        # try to infer label from any obvious column names
        for cand in ['benign_malignant','pathology','target','class','status']:
            if cand in cols:
                out['label'] = df[cols[cand]].apply(coerce_label)
                break

    # drop rows without label
    if 'label' not in out.columns:
        print(f"  ⚠️ {pcsv.name}: no label-like column found.")
        return None
    out = out.dropna(subset=['label']).copy()
    out['label'] = out['label'].astype(int).clip(0,1)

    # normalize any path column to absolute path under DATA_ROOT (use basename matching)
    if 'path' in out.columns:
        # strip quotes & slashes
        paths = []
        for s in out['path'].astype(str).tolist():
            s2 = s.strip().strip('"').strip("'").replace('\\','/')
            cand = Path(s2)
            if cand.is_absolute() and cand.exists():
                paths.append(str(cand.resolve()))
            else:
                # try resolve relative to DATA_ROOT or match by basename against dicom_df
                name = Path(s2).name
                hit = dicom_df[dicom_df['path'].str.endswith('/'+name, na=False)]
                if len(hit) >= 1:
                    paths.append(hit['path'].iloc[0])
                else:
                    guess = DATA_ROOT / s2
                    paths.append(str(guess.resolve()) if guess.exists() else None)
        out['path'] = paths
    return out

label_frames = []
for pcsv in csv_candidates:
    lf = read_label_csv(pcsv)
    if lf is not None and len(lf):
        print(f"  ✓ {pcsv.name}: {len(lf)} rows with labels")
        label_frames.append(lf)

if not label_frames:
    print("⚠️ No usable CSV with labels found; will try folder-name heuristics.")
    # Heuristic: label from folder names 0/1 or benign/malignant
    rows = []
    for p in dicom_df['path']:
        s = p.lower()
        lab = (1 if any(tok in s for tok in ['malignant','malign','cancer','/1/','\\1\\','_1','-1'])
               else 0 if any(tok in s for tok in ['benign','/0/','\\0\\','_0','-0'])
               else None)
        if lab is not None:
            rows.append((p, lab))
    df_all = pd.DataFrame(rows, columns=['path','label'])
else:
    labels_all = pd.concat(label_frames, ignore_index=True).drop_duplicates().reset_index(drop=True)

    # Join priority: SOPInstanceUID -> SeriesInstanceUID -> normalized path
    df_all = pd.DataFrame(columns=['path','label'])

    if 'SOPInstanceUID' in labels_all.columns and labels_all['SOPInstanceUID'].notna().any():
        left = labels_all.dropna(subset=['SOPInstanceUID']).copy()
        j = left.merge(dicom_df[['path','SOPInstanceUID']], on='SOPInstanceUID', how='left')
        j = j.dropna(subset=['path','label'])
        df_all = pd.concat([df_all, j[['path','label']]], ignore_index=True)

    if 'SeriesInstanceUID' in labels_all.columns and labels_all['SeriesInstanceUID'].notna().any():
        left = labels_all.dropna(subset=['SeriesInstanceUID']).copy()
        j = left.merge(dicom_df[['path','SeriesInstanceUID']], on='SeriesInstanceUID', how='left')
        j = j.dropna(subset=['path','label'])
        df_all = pd.concat([df_all, j[['path','label']]], ignore_index=True)

    if 'path' in labels_all.columns and labels_all['path'].notna().any():
        left = labels_all.dropna(subset=['path']).copy()
        # If the path column is absolute and exists, keep; else try to resolve by basename
        fixed = []
        for s in left['path'].astype(str).tolist():
            s2 = s.strip().strip('"').strip("'").replace('\\','/')
            cand = Path(s2)
            if cand.exists():
                fixed.append(str(cand.resolve()))
            else:
                # match by basename in dicom_df
                name = Path(s2).name
                hit = dicom_df[dicom_df['path'].str.endswith('/'+name, na=False)]
                fixed.append(hit['path'].iloc[0] if len(hit) else None)
        left['path'] = fixed
        left = left.dropna(subset=['path'])
        df_all = pd.concat([df_all, left[['path','label']]], ignore_index=True)

    # de-dup, keep first occurrence
    df_all = df_all.dropna(subset=['path','label']).drop_duplicates('path').reset_index(drop=True)

# Final sanity
df_all['label'] = df_all['label'].astype(int).clip(0,1)
df_all = df_all[df_all['path'].apply(lambda s: Path(s).exists())].reset_index(drop=True)

print(f"✅ df_all built: {len(df_all)} rows")
print("Label counts:", df_all['label'].value_counts().to_dict())
print(df_all.head(5))

# Save for reuse
CSV_OUT = OUT_DIR / 'train_paths_labels.normalized.csv'
df_all.to_csv(CSV_OUT, index=False)
print("Wrote:", CSV_OUT)
```

```python
# %% [colab] PATCH: robust join -> build df_all without KeyError
import pandas as pd
from pathlib import Path

def resolve_to_existing_path(s):
    if s is None or (isinstance(s, float) and pd.isna(s)):
        return None
    s2 = str(s).strip().strip('"').strip("'").replace('\\','/')
    cand = Path(s2)
    if cand.exists():
        return str(cand.resolve())
    # try basename match inside dicom_df
    name = Path(s2).name
    hit = dicom_df[dicom_df['path'].str.endswith('/'+name, na=False)]
    if len(hit):
        return hit['path'].iloc[0]
    guess = (DATA_ROOT / name).resolve()
    return str(guess) if guess.exists() else None

def pick_path(row):
    # prefer DICOM-resolved path, else try original CSV path normalized
    pdcm = row.get('path_dcm', None)
    if pdcm is not None and isinstance(pdcm, str) and len(pdcm):
        return pdcm
    return resolve_to_existing_path(row.get('path', None))

df_all = pd.DataFrame(columns=['path','label'])

# 1) SOPInstanceUID → path
if 'SOPInstanceUID' in labels_all.columns:
    left = labels_all.dropna(subset=['SOPInstanceUID']).copy()
    if len(left):
        j = left.merge(
            dicom_df[['SOPInstanceUID','path']].rename(columns={'path':'path_dcm'}),
            on='SOPInstanceUID', how='left'
        )
        j['path'] = j.apply(pick_path, axis=1)
        j = j.dropna(subset=['path','label'])
        df_all = pd.concat([df_all, j[['path','label']]], ignore_index=True)

# 2) SeriesInstanceUID → path
if 'SeriesInstanceUID' in labels_all.columns:
    left = labels_all.dropna(subset=['SeriesInstanceUID']).copy()
    if len(left):
        j = left.merge(
            dicom_df[['SeriesInstanceUID','path']].rename(columns={'path':'path_dcm'}),
            on='SeriesInstanceUID', how='left'
        )
        j['path'] = j.apply(pick_path, axis=1)
        j = j.dropna(subset=['path','label'])
        df_all = pd.concat([df_all, j[['path','label']]], ignore_index=True)

# 3) Path-only rows from CSVs (normalize to absolute / match by basename)
if 'path' in labels_all.columns:
    left = labels_all.dropna(subset=['path']).copy()
    if len(left):
        left['path'] = left['path'].apply(resolve_to_existing_path)
        left = left.dropna(subset=['path'])
        df_all = pd.concat([df_all, left[['path','label']]], ignore_index=True)

# Final clean-up
df_all = df_all.drop_duplicates('path').reset_index(drop=True)
df_all = df_all[df_all['path'].apply(lambda s: Path(s).exists())].reset_index(drop=True)
df_all['label'] = df_all['label'].astype(int).clip(0,1)

print(f"✅ df_all built: {len(df_all)} rows")
print("Label counts:", df_all['label'].value_counts().to_dict())
print(df_all.head(5))

CSV_OUT = OUT_DIR / 'train_paths_labels.normalized.csv'
df_all.to_csv(CSV_OUT, index=False)
print("Wrote:", CSV_OUT)
```

```python
# %% [colab] Med-SA Adapter — 80/10/10 split → train → test
!pip -q install --no-input timm==1.0.9 scikit-learn albumentations pydicom opencv-python

import os, sys, random, numpy as np, pandas as pd, torch, torch.nn as nn
import cv2, pydicom, timm
from pathlib import Path
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.metrics import roc_auc_score, accuracy_score, confusion_matrix

assert 'df_all' in globals() and len(df_all)>0, "Run Cell 1; df_all is empty."
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
SEED, EPOCHS, LR, WD = 42, 20, 2e-4, 1e-2
IMG_SIZE, BATCH = 224, 16
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)

# ----- split (exact 80/10/10 best-effort stratified) -----
df_all = df_all.dropna(subset=['path','label']).copy()
df_all['label'] = df_all['label'].astype(int).clip(0,1)
n = len(df_all)
print(f"Total={n} | label counts:", df_all['label'].value_counts().to_dict())

n_test = max(1, int(round(0.10*n))) if n>=10 else max(0, min(1, n-1))
n_val  = max(1, int(round(0.10*n))) if n>=10 else max(0, min(1, n-1-n_test))
n_train = max(1, n - n_test - n_val)

y = df_all['label'].to_numpy()
can_strat = (np.unique(y).size == 2) and (df_all['label'].value_counts().min() >= 2)

idx_all = np.arange(n)
if n_test>0 and can_strat and n_test>=2:
    sss = StratifiedShuffleSplit(n_splits=1, test_size=n_test, random_state=SEED)
    idx_trval, idx_test = next(sss.split(np.zeros(n), y))
else:
    rng = np.random.RandomState(SEED); rng.shuffle(idx_all)
    idx_test = idx_all[:n_test]; idx_trval = idx_all[n_test:]

y_tv = y[idx_trval]
can_strat_tv = (np.unique(y_tv).size == 2) and (pd.Series(y_tv).value_counts().min() >= 1)
if n_val>0 and can_strat_tv and n_val>=2:
    sss2 = StratifiedShuffleSplit(n_splits=1, test_size=n_val, random_state=SEED+1)
    dummy = np.zeros(len(idx_trval))
    it_tr, it_val = next(sss2.split(dummy, y_tv))
    idx_val = idx_trval[it_val]; idx_train = idx_trval[it_tr]
else:
    rng = np.random.RandomState(SEED+1)
    shuf = idx_trval.copy(); rng.shuffle(shuf)
    idx_val = shuf[:n_val]; idx_train = shuf[n_val:]

used = set(idx_train.tolist()+idx_val.tolist()+idx_test.tolist())
if len(used) != n:
    leftovers = list(set(range(n)) - used)
    idx_train = np.concatenate([idx_train, np.array(leftovers, dtype=int)])

train_df = df_all.iloc[idx_train].reset_index(drop=True)
val_df   = df_all.iloc[idx_val].reset_index(drop=True)
test_df  = df_all.iloc[idx_test].reset_index(drop=True)
print(f"Splits -> train={len(train_df)}, val={len(val_df)}, test={len(test_df)} (sum={len(train_df)+len(val_df)+len(test_df)})")

# ----- transforms -----
import albumentations as A
from albumentations.pytorch import ToTensorV2

train_tf = A.Compose([A.Resize(IMG_SIZE, IMG_SIZE), A.HorizontalFlip(p=0.5), A.RandomBrightnessContrast(p=0.2), ToTensorV2()])
eval_tf  = A.Compose([A.Resize(IMG_SIZE, IMG_SIZE), ToTensorV2()])

# ----- IO & dataset -----
def read_image_any(p: Path) -> np.ndarray:
    p = Path(p)
    if p.suffix.lower()=='.dcm' or True:  # most are DICOM
        try:
            ds = pydicom.dcmread(str(p))
            img = ds.pixel_array.astype(np.float32)
            img -= img.min()
            if img.max() > 0: img /= img.max()
            img = (img*255.0).clip(0,255).astype(np.uint8)
        except Exception:
            # fall back to cv2 if it's actually PNG/JPG
            img = cv2.imread(str(p), cv2.IMREAD_GRAYSCALE)
            if img is None: raise FileNotFoundError(p)
    else:
        img = cv2.imread(str(p), cv2.IMREAD_GRAYSCALE)
        if img is None: raise FileNotFoundError(p)
    return img

class MammoAdapterCls(torch.utils.data.Dataset):
    def __init__(self, df, tfm):
        self.paths = df['path'].tolist()
        self.labels = df['label'].astype(np.float32).tolist()
        self.tfm = tfm
    def __len__(self): return len(self.paths)
    def __getitem__(self, i):
        p = Path(self.paths[i]); y = self.labels[i]
        img = read_image_any(p)
        if img.ndim==2: img = np.stack([img,img,img], axis=-1)
        x = self.tfm(image=img)['image'].float()
        return x, torch.tensor(y), str(p)

from torch.utils.data import DataLoader
def fit_bs(default_bs, n): return max(1, min(default_bs, n if n>0 else 1))
pin = torch.cuda.is_available()
train_ds = MammoAdapterCls(train_df, train_tf)
val_ds   = MammoAdapterCls(val_df,   eval_tf)
test_ds  = MammoAdapterCls(test_df,  eval_tf)
BATCH = fit_bs(BATCH, len(train_ds))
train_loader = DataLoader(train_ds, batch_size=BATCH,                        shuffle=True,  num_workers=2, pin_memory=pin)
val_loader   = DataLoader(val_ds,   batch_size=fit_bs(BATCH*2, len(val_ds)), shuffle=False, num_workers=2, pin_memory=pin)
test_loader  = DataLoader(test_ds,  batch_size=fit_bs(BATCH*2, len(test_ds)), shuffle=False, num_workers=2, pin_memory=pin)

# ----- model: Med-SA–style adapter on frozen ViT-S/16 -----
class BottleneckAdapter(nn.Module):
    def __init__(self, dim, bottleneck=64, p_drop=0.0):
        super().__init__()
        self.down = nn.Linear(dim, bottleneck); self.act=nn.GELU()
        self.up   = nn.Linear(bottleneck, dim); self.drop=nn.Dropout(p_drop)
    def forward(self, x): return x + self.drop(self.up(self.act(self.down(x))))

class AdapterClassifier(nn.Module):
    def __init__(self, backbone='vit_small_patch16_224', adapter_bottleneck=64, dropout=0.1):
        super().__init__()
        self.backbone = timm.create_model(backbone, pretrained=True, num_classes=0, in_chans=3)
        dim = self.backbone.num_features
        for p in self.backbone.parameters(): p.requires_grad=False
        self.adapter1 = BottleneckAdapter(dim, adapter_bottleneck, dropout)
        self.adapter2 = BottleneckAdapter(dim, adapter_bottleneck, dropout)
        self.head = nn.Linear(dim, 1)
    def forward(self, x255):
        x = x255/255.0
        z = self.backbone(x); z=self.adapter2(self.adapter1(z))
        return self.head(z).squeeze(1)

model = AdapterClassifier().to(DEVICE)

# ----- loss/opt/amp -----
pos = float((train_df['label']==1).sum()); neg = float((train_df['label']==0).sum())
pos_weight_val = 1.0 if (pos==0 or neg==0) else (neg/max(pos,1.0))
criterion = nn.BCEWithLogitsLoss(pos_weight=torch.tensor([pos_weight_val]).to(DEVICE))
OPT = torch.optim.AdamW((p for p in model.parameters() if p.requires_grad), lr=LR, weight_decay=WD)
SCALER = torch.amp.GradScaler('cuda', enabled=torch.cuda.is_available())

BEST_CKPT = Path('/content/drive/MyDrive/checkpoints/medsa_style_cls_best_80_10_10.pth')
BEST_CKPT.parent.mkdir(parents=True, exist_ok=True)

def evaluate(loader):
    if loader is None or len(loader.dataset)==0:
        return dict(N=0, acc=np.nan, auc=np.nan, rows=[])
    model.eval(); ys, ps, rows = [], [], []
    with torch.no_grad():
        for x,y,paths in loader:
            x = x.to(DEVICE, non_blocking=True); y = y.to(DEVICE, non_blocking=True)
            with torch.amp.autocast('cuda', enabled=torch.cuda.is_available()):
                prob = torch.sigmoid(model(x))
            ys.append(y.cpu().numpy()); ps.append(prob.cpu().numpy())
            for pth, yy, pp in zip(paths, y.cpu().numpy(), prob.cpu().numpy()):
                rows.append(dict(path=pth, label=int(yy), prob=float(pp), pred=int(pp>=0.5)))
    ys = np.concatenate(ys) if ys else np.array([])
    ps = np.concatenate(ps) if ps else np.array([])
    pred = (ps>=0.5).astype(np.float32) if len(ps) else np.array([])
    acc = accuracy_score(ys, pred) if len(ys) else np.nan
    try: auc = roc_auc_score(ys, ps)
    except Exception: auc = np.nan
    return dict(N=len(ys), acc=acc, auc=auc, rows=rows)

# ----- train -----
best_metric, best_value = None, -1.0
for epoch in range(1, EPOCHS+1):
    model.train(); running=0.0
    for step, (x,y,_) in enumerate(train_loader, 1):
        x = x.to(DEVICE, non_blocking=True); y = y.to(DEVICE, non_blocking=True)
        OPT.zero_grad(set_to_none=True)
        with torch.amp.autocast('cuda', enabled=torch.cuda.is_available()):
            loss = criterion(model(x), y)
        SCALER.scale(loss).backward(); SCALER.step(OPT); SCALER.update()
        running += float(loss.detach())
        if step % 50 == 0 or step==len(train_loader):
            print(f"Epoch {epoch} step {step}: loss={running/step:.4f}")

    val_stats = evaluate(val_loader)
    print(f"Epoch {epoch} — val N={val_stats['N']}  Acc={val_stats['acc']:.4f}  AUC={val_stats['auc']:.4f}")

    if val_stats['N']>0:
        metric_name  = 'auc' if not np.isnan(val_stats['auc']) else 'acc'
        metric_value = val_stats[metric_name]
        if (best_metric is None or
            (metric_name == 'auc' and best_metric == 'acc') or
            (metric_name == best_metric and metric_value > best_value)):
            best_metric, best_value = metric_name, metric_value
            torch.save({'model': model.state_dict(),
                        'epoch': epoch,
                        'val_acc': val_stats['acc'],
                        'val_auc': val_stats['auc'],
                        'metric': best_metric,
                        'metric_value': best_value}, BEST_CKPT)
            print(f"✅ Saved best ({best_metric.upper()}={best_value:.4f}) -> {BEST_CKPT}")
    else:
        torch.save({'model': model.state_dict(), 'epoch': epoch}, BEST_CKPT)
        print(f"💾 Saved (no-val) epoch {epoch} -> {BEST_CKPT}")

# ----- test -----
if BEST_CKPT.exists():
    ckpt = torch.load(BEST_CKPT, map_location='cpu')
    model.load_state_dict(ckpt['model'])
    print(f"Loaded checkpoint from epoch {ckpt.get('epoch','?')}")

train_stats = evaluate(train_loader)
val_stats   = evaluate(val_loader)
test_stats  = evaluate(test_loader)

import pandas as pd
summary = pd.DataFrame([
    dict(Split='Train', N=train_stats['N'], Accuracy=None if np.isnan(train_stats['acc']) else round(train_stats['acc'],4),
         ROC_AUC=None if np.isnan(train_stats['auc']) else round(train_stats['auc'],4)),
    dict(Split='Val',   N=val_stats['N'],   Accuracy=None if np.isnan(val_stats['acc']) else round(val_stats['acc'],4),
         ROC_AUC=None if np.isnan(val_stats['auc']) else round(val_stats['auc'],4)),
    dict(Split='Test',  N=test_stats['N'],  Accuracy=None if np.isnan(test_stats['acc']) else round(test_stats['acc'],4),
         ROC_AUC=None if np.isnan(test_stats['auc']) else round(test_stats['auc'],4)),
])
print("\n=== Summary ===")
print(summary.to_string(index=False))

OUT_DIR = Path('/content/drive/MyDrive/cbis_cls_runs'); OUT_DIR.mkdir(parents=True, exist_ok=True)
if test_stats['N']>0:
    pd.DataFrame(test_stats['rows']).to_csv(OUT_DIR/'preds_test.csv', index=False)
else:
    pd.DataFrame(val_stats['rows']).to_csv(OUT_DIR/'preds_val_as_test.csv', index=False)
print(f"Saved per-image predictions to: {OUT_DIR}")
```
